```python
import pandas as pd
import numpy as np

data=pd.read_csv('https://raw.githubusercontent.com/laxmimerit/All-CSV-ML-Data-Files-Download/master/IMDB-Movie-Data.csv')
data
```

```python
print(f'Number of duplicate rows: {data.duplicated().sum()}')
data.drop_duplicates(inplace=True)
```

```python
data.rename(columns={'Title': 'Movie_Title'}, inplace=True)
data.columns
```

```python
unique_genres = data['Genre'].str.split(',').explode().unique()
print(f'Total number of unique genres: {len(unique_genres)}')
print('Unique genres:', unique_genres)
```

```python
data['Number_of_Genres'] = data['Genre'].apply(lambda x: len(str(x).split(',')))
print(data[['Movie_Title', 'Genre', 'Number_of_Genres']].head())
```

```python
data['Revenue (Millions)'].fillna(data['Revenue (Millions)'].median(), inplace=True)

data['Metascore'].fillna(data['Metascore'].median(), inplace=True)

print(data.isnull().sum())
```

```python
top_5_movies_by_rating = data.sort_values(by='Rating', ascending=False).head(5)
print('Top 5 movies by Rating:')
print(top_5_movies_by_rating[['Movie_Title', 'Rating']])
```

```python
highest_revenue_movie = data.loc[data['Revenue (Millions)'].idxmax()]
lowest_revenue_movie = data.loc[data['Revenue (Millions)'].idxmin()]

print('Movie with Highest Revenue:')
print(highest_revenue_movie[['Movie_Title', 'Revenue (Millions)']])

print('\nMovie with Lowest Revenue:')
print(lowest_revenue_movie[['Movie_Title', 'Revenue (Millions)']])
```

```python
avg_rating_by_director = data.groupby('Director')['Rating'].mean().sort_values(ascending=False)
print('Average Rating by Director (Top 10):')
print(avg_rating_by_director.head(10))
```

```python
movies_per_year = data['Year'].value_counts().sort_index()
print('Number of Movies per Year:')
print(movies_per_year)
```

```python
high_rating_high_revenue_movies = data[(data['Rating'] > 8) & (data['Revenue (Millions)'] > 100)]
print('Movies with Rating > 8 and Revenue > 100M:')
print(high_rating_high_revenue_movies[['Movie_Title', 'Rating', 'Revenue (Millions)']])
```

```python
director_stats = data.groupby('Director').agg({'Rating': 'mean', 'Revenue (Millions)': 'mean'})
print('Average Rating and Revenue by Director:')
print(director_stats.sort_values(by='Rating', ascending=False).head())
```

```python
year_stats = data.groupby('Year').agg(Count=('Movie_Title', 'count'), Average_Rating=('Rating', 'mean'))
print('Number of Movies and Average Rating by Year:')
print(year_stats)
```

```python
data['Profitability'] = data['Revenue (Millions)'] / data['Runtime (Minutes)']
most_profitable_movies = data.sort_values(by='Profitability', ascending=False).head(10)
print('Top 10 Most Profitable Movies (Revenue per Minute):')
print(most_profitable_movies[['Movie_Title', 'Revenue (Millions)', 'Runtime (Minutes)', 'Profitability']])
```

```python
data['Rating_Category'] = data['Rating'].apply(
    lambda x: 'Poor' if x < 5.9
    else ('Average' if x < 6.9
          else ('Good' if x < 7.9 else 'Excellent'))
)

print('Distribution of movies by Rating Category:')
print(data['Rating_Category'].value_counts())
```

```python
def custom_movie_score(row):
  High_score = row['Rating'] * 10 + row['Revenue (Millions)']
  return High_score

data['Movie_Score'] = data.apply(custom_movie_score, axis=1)
print('Top 10 Movies by Custom Score:')
print(data.sort_values(by='Movie_Score', ascending=False).head(10)[['Movie_Title', 'Rating', 'Revenue (Millions)', 'Metascore', 'Movie_Score']])
```
